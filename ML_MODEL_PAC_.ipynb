```python

# ONE-CELL COMPLETE ML + SHAP + PUMICE REPLACEMENT ERROR ANALYSIS
# PAC COMPRESSIVE STRENGTH PREDICTION
#
# This single Google Colab script:
# 1. Uploads ONE or MORE CSV files and combines them.
# 2. Cleans duplicates/incomplete rows and detects N dynamically.
# 3. Runs shuffled 10-fold CV for RF, GBR, XGB and SVR.
# 4. Reports mean ± SD for R², RMSE, MAE and MAPE.
# 5. Saves out-of-fold predictions and observed-vs-predicted plots.
# 6. Automatically selects the model with the highest mean CV R².
# 7. Fits that selected model to the complete cleaned dataset ONLY
#    for SHAP interpretation.
# 8. Runs SHAP importance/dependence and tree-SHAP interactions.
# 9. Performs pumice-replacement-range error analysis using OOF
#    predictions only.
#
# Expected columns:
# Cement, Pumice_Aggregate, Natural_Aggregate, Sand, Water,
# Compressive_Strength
#
# No fixed 131-observation requirement.
# ================================================================

!pip -q install shap xgboost openpyxl

import os
import glob
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.model_selection import KFold
from sklearn.base import clone
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

from xgboost import XGBRegressor
import shap

warnings.filterwarnings("ignore")

RANDOM_STATE = 42
N_SPLITS = 10
FIG_DPI = 600

# For very large databases SHAP can be slow. All rows are used up to
# this limit; the limit does NOT affect CV performance.
MAX_SHAP_OBSERVATIONS = 500
MAX_SHAP_BACKGROUND = 100

print("============================================================")
print("PAC COMPLETE ML + SHAP + ERROR ANALYSIS")
print("============================================================")


# ================================================================
# 1. UPLOAD ONE OR MORE CSV FILES
# ================================================================

print("\nUpload one or more CSV files.")
print("Multiple CSVs will be combined row-wise.")
uploaded = files.upload()

if not uploaded:
    raise ValueError("No CSV file was uploaded.")

dataframes = []

for file_name in uploaded.keys():

    if not file_name.lower().endswith(".csv"):
        print(f"Skipping non-CSV file: {file_name}")
        continue

    temp = pd.read_csv(file_name)

    print(
        f"Loaded: {file_name} | "
        f"rows = {len(temp)}, columns = {len(temp.columns)}"
    )

    dataframes.append(temp)

if not dataframes:
    raise ValueError("No CSV files were uploaded.")

raw_df = pd.concat(
    dataframes,
    axis=0,
    ignore_index=True
)

print("\nCombined raw dataset shape:", raw_df.shape)


# ================================================================
# 2. REQUIRED COLUMNS
# ================================================================

feature_columns = [
    "Cement",
    "Pumice_Aggregate",
    "Natural_Aggregate",
    "Sand",
    "Water"
]

target_column = "Compressive_Strength"

required_columns = feature_columns + [target_column]

missing_columns = [
    col for col in required_columns
    if col not in raw_df.columns
]

if missing_columns:
    raise ValueError(
        "Missing required columns:\n"
        + "\n".join([f" - {x}" for x in missing_columns])
        + "\n\nColumns found:\n"
        + "\n".join([f" - {x}" for x in raw_df.columns])
    )

print("\n✓ All required columns found.")


# ================================================================
# 3. SELECT MODEL COLUMNS + CONVERT TO NUMERIC
# ================================================================

df = raw_df[required_columns].copy()

for col in required_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ================================================================
# 4. REMOVE EXACT DUPLICATES
# ================================================================

duplicate_count = df.duplicated().sum()

print("\nExact duplicate rows:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)


# ================================================================
# 5. REMOVE INCOMPLETE / INVALID RECORDS
# ================================================================

print("\nMissing values before cleaning:")
print(df.isna().sum())

incomplete_rows = df.isna().any(axis=1).sum()

print("\nIncomplete rows:", incomplete_rows)

df = (
    df
    .dropna(subset=required_columns)
    .reset_index(drop=True)
)

if np.isinf(
    df[required_columns].to_numpy(dtype=float)
).any():
    raise ValueError(
        "Infinite values were detected in the model columns."
    )

if len(df) < N_SPLITS:
    raise ValueError(
        f"Final dataset has {len(df)} observations. "
        f"At least {N_SPLITS} are required for {N_SPLITS}-fold CV."
    )

print("\n============================================================")
print("FINAL DATASET")
print("============================================================")
print("Final observations:", len(df))
print(
    "Compressive strength range:",
    f"{df[target_column].min():.3f} to "
    f"{df[target_column].max():.3f} MPa"
)


# ================================================================
# 6. X AND y
# ================================================================

X = df[feature_columns].copy()
y = df[target_column].copy()


# ================================================================
# 7. FINAL MODELS / HYPERPARAMETERS
# ================================================================
# These are the same final parameters used in the current analysis.
#
# SVR scaling is inside its Pipeline, so the scaler is fitted only
# on each training fold and never on the held-out fold.
# ================================================================

models = {

    "RF": RandomForestRegressor(
        n_estimators=300,
        max_depth=10,
        min_samples_split=2,
        min_samples_leaf=1,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "GBR": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.10,
        max_depth=3,
        min_samples_split=2,
        min_samples_leaf=2,
        random_state=RANDOM_STATE
    ),

    "XGB": XGBRegressor(
        n_estimators=200,
        learning_rate=0.10,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=RANDOM_STATE,
        n_jobs=-1,
        verbosity=0
    ),

    "SVR": Pipeline([
        ("scaler", StandardScaler()),
        ("svr", SVR(
            kernel="rbf",
            C=100,
            gamma="scale",
            epsilon=0.01
        ))
    ])
}


# ================================================================
# 8. CV SETUP
# ================================================================

cv = KFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)


# ================================================================
# 9. METRICS
# ================================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    r2 = r2_score(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    nonzero = y_true != 0

    if np.any(nonzero):
        mape = (
            np.mean(
                np.abs(
                    (
                        y_true[nonzero]
                        - y_pred[nonzero]
                    )
                    /
                    y_true[nonzero]
                )
            )
            * 100
        )
    else:
        mape = np.nan

    return r2, rmse, mae, mape


# ================================================================
# 10. 10-FOLD CROSS-VALIDATION + OOF PREDICTIONS
# ================================================================

fold_results = []
oof_predictions = {}

for model_name, model in models.items():

    print(f"\nRunning {N_SPLITS}-fold CV for {model_name}...")

    y_pred_oof = np.zeros(
        len(y),
        dtype=float
    )

    for fold_number, (train_idx, test_idx) in enumerate(
        cv.split(X),
        start=1
    ):

        X_train = X.iloc[train_idx]
        X_test = X.iloc[test_idx]

        y_train = y.iloc[train_idx]
        y_test = y.iloc[test_idx]

        fold_model = clone(model)

        # Training only on training fold
        fold_model.fit(
            X_train,
            y_train
        )

        # Held-out prediction
        y_pred = fold_model.predict(
            X_test
        )

        y_pred_oof[test_idx] = y_pred

        r2, rmse, mae, mape = calculate_metrics(
            y_test,
            y_pred
        )

        fold_results.append({
            "Model": model_name,
            "Fold": fold_number,
            "R2": r2,
            "RMSE_MPa": rmse,
            "MAE_MPa": mae,
            "MAPE_percent": mape
        })

    oof_predictions[model_name] = y_pred_oof

    print(f"✓ {model_name} completed.")


fold_results_df = pd.DataFrame(
    fold_results
)


# ================================================================
# 11. MEAN ± SD
# ================================================================

summary_rows = []

for model_name in models.keys():

    r = fold_results_df[
        fold_results_df["Model"] == model_name
    ]

    summary_rows.append({

        "Model": model_name,

        "R2_mean": r["R2"].mean(),
        "R2_SD": r["R2"].std(ddof=1),

        "RMSE_mean": r["RMSE_MPa"].mean(),
        "RMSE_SD": r["RMSE_MPa"].std(ddof=1),

        "MAE_mean": r["MAE_MPa"].mean(),
        "MAE_SD": r["MAE_MPa"].std(ddof=1),

        "MAPE_mean": r["MAPE_percent"].mean(),
        "MAPE_SD": r["MAPE_percent"].std(ddof=1)
    })

summary_df = pd.DataFrame(
    summary_rows
)


# ================================================================
# 12. SELECT MODEL FOR SHAP
# ================================================================
# Selection criterion is explicitly defined as highest mean 10-fold
# CV R². This is model selection for interpretation, not a universal
# ranking of ML algorithms.
# ================================================================

best_row = (
    summary_df
    .sort_values(
        "R2_mean",
        ascending=False
    )
    .iloc[0]
)

best_model_name = best_row["Model"]

print("\n============================================================")
print("MODEL PERFORMANCE")
print("============================================================")

for _, row in summary_df.iterrows():

    print(f"\n{row['Model']}")

    print(
        f"R²    = {row['R2_mean']:.3f} ± "
        f"{row['R2_SD']:.3f}"
    )

    print(
        f"RMSE  = {row['RMSE_mean']:.3f} ± "
        f"{row['RMSE_SD']:.3f} MPa"
    )

    print(
        f"MAE   = {row['MAE_mean']:.3f} ± "
        f"{row['MAE_SD']:.3f} MPa"
    )

    print(
        f"MAPE  = {row['MAPE_mean']:.2f} ± "
        f"{row['MAPE_SD']:.2f}%"
    )

print("\nModel selected for SHAP:", best_model_name)


# ================================================================
# 13. SAVE PERFORMANCE TABLES
# ================================================================

table5_publication = pd.DataFrame({

    "Model": summary_df["Model"],

    "R²": (
        summary_df["R2_mean"].map(lambda x: f"{x:.3f}")
        + " ± "
        + summary_df["R2_SD"].map(lambda x: f"{x:.3f}")
    ),

    "RMSE (MPa)": (
        summary_df["RMSE_mean"].map(lambda x: f"{x:.3f}")
        + " ± "
        + summary_df["RMSE_SD"].map(lambda x: f"{x:.3f}")
    ),

    "MAE (MPa)": (
        summary_df["MAE_mean"].map(lambda x: f"{x:.3f}")
        + " ± "
        + summary_df["MAE_SD"].map(lambda x: f"{x:.3f}")
    ),

    "MAPE (%)": (
        summary_df["MAPE_mean"].map(lambda x: f"{x:.2f}")
        + " ± "
        + summary_df["MAPE_SD"].map(lambda x: f"{x:.2f}")
    )
})

print("\n============================================================")
print("TABLE 5")
print("============================================================")
display(table5_publication)

fold_results_df.to_csv(
    "Fold_by_Fold_CV_Results.csv",
    index=False
)

summary_df.to_csv(
    "Table_5_Full_Numerical_Results.csv",
    index=False
)

table5_publication.to_csv(
    "Table_5_Publication_Format.csv",
    index=False
)


# ================================================================
# 14. SAVE OOF PREDICTIONS
# ================================================================

oof_df = df.copy()

for model_name, predictions in oof_predictions.items():
    oof_df[
        model_name + "_Prediction"
    ] = predictions

oof_df.to_csv(
    "Out_of_Fold_Predictions_All_Models.csv",
    index=False
)


# ================================================================
# 15. OBSERVED VS PREDICTED FIGURE
# ================================================================

all_values = np.concatenate(
    [y.to_numpy()]
    +
    list(oof_predictions.values())
)

data_min = np.min(all_values)
data_max = np.max(all_values)

margin = 0.05 * (data_max - data_min)

if margin == 0:
    margin = 1.0

axis_min = max(
    0,
    data_min - margin
)

axis_max = data_max + margin

x_line = np.linspace(
    axis_min,
    axis_max,
    300
)

fig, axes = plt.subplots(
    2,
    2,
    figsize=(7.2, 6.4),
    dpi=FIG_DPI
)

axes = axes.flatten()

plot_order = [
    ("RF", "Random Forest"),
    ("GBR", "Gradient Boosting Regressor"),
    ("XGB", "XGBoost"),
    ("SVR", "Support Vector Regression")
]

panel_labels = [
    "(a)",
    "(b)",
    "(c)",
    "(d)"
]

for ax, (model_name, model_label), panel in zip(
    axes,
    plot_order,
    panel_labels
):

    y_pred = oof_predictions[model_name]

    ax.scatter(
        y,
        y_pred,
        s=28,
        alpha=0.75,
        edgecolors="black",
        linewidths=0.35
    )

    ax.plot(
        x_line,
        x_line,
        linestyle="-",
        linewidth=1.8
    )

    ax.plot(
        x_line,
        1.20 * x_line,
        linestyle="--",
        linewidth=1.2
    )

    ax.plot(
        x_line,
        0.80 * x_line,
        linestyle="--",
        linewidth=1.2
    )

    mean_r2 = summary_df.loc[
        summary_df["Model"] == model_name,
        "R2_mean"
    ].iloc[0]

    ax.text(
        0.04,
        0.95,
        panel,
        transform=ax.transAxes,
        fontsize=12,
        fontweight="bold",
        va="top"
    )

    ax.text(
        0.96,
        0.95,
        model_label,
        transform=ax.transAxes,
        fontsize=9.5,
        fontweight="bold",
        ha="right",
        va="top"
    )

    ax.text(
        0.04,
        0.86,
        f"$R^2$ = {mean_r2:.3f}",
        transform=ax.transAxes,
        fontsize=9.5,
        va="top"
    )

    ax.set_xlim(axis_min, axis_max)
    ax.set_ylim(axis_min, axis_max)

    ax.set_xlabel(
        "Observed compressive strength (MPa)",
        fontsize=10
    )

    ax.set_ylabel(
        "Predicted compressive strength (MPa)",
        fontsize=10
    )

    ax.tick_params(
        axis="both",
        which="major",
        labelsize=9,
        direction="out",
        length=4,
        width=0.8
    )

    for spine in ax.spines.values():
        spine.set_linewidth(0.9)

    ax.grid(False)

plt.tight_layout(
    pad=1.2,
    w_pad=1.2,
    h_pad=1.2
)

plt.savefig(
    "Figure_Observed_vs_Predicted_600dpi.tiff",
    dpi=FIG_DPI,
    format="tiff",
    bbox_inches="tight"
)

plt.savefig(
    "Figure_Observed_vs_Predicted_600dpi.png",
    dpi=FIG_DPI,
    format="png",
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 16. PUMICE REPLACEMENT PERCENTAGE
# ================================================================

aggregate_total = (
    df["Pumice_Aggregate"]
    +
    df["Natural_Aggregate"]
)

if (aggregate_total <= 0).any():
    raise ValueError(
        "One or more observations have zero/negative total "
        "coarse aggregate."
    )

df["Pumice_Replacement_Percent"] = (
    df["Pumice_Aggregate"]
    /
    aggregate_total
) * 100


def classify_replacement(p):

    if np.isclose(p, 0.0, atol=1e-8):
        return "0%"

    elif np.isclose(p, 100.0, atol=1e-8):
        return "100%"

    elif p < 50.0:
        return ">0–<50%"

    else:
        return "≥50–<100%"


df["Replacement_Group"] = (
    df["Pumice_Replacement_Percent"]
    .apply(classify_replacement)
)

group_order = [
    "0%",
    ">0–<50%",
    "≥50–<100%",
    "100%"
]

group_counts = (
    df["Replacement_Group"]
    .value_counts()
    .reindex(group_order)
    .fillna(0)
    .astype(int)
)

print("\n============================================================")
print("PUMICE REPLACEMENT DISTRIBUTION")
print("============================================================")

for group, n in group_counts.items():
    print(f"{group:12s}: n = {n}")


# ================================================================
# 17. ERROR ANALYSIS BY REPLACEMENT RANGE
# ================================================================
# Uses OOF predictions only.
# ================================================================

range_results = []

for model_name, predictions in oof_predictions.items():

    for group in group_order:

        mask = (
            df["Replacement_Group"] == group
        )

        y_group = df.loc[
            mask,
            target_column
        ]

        pred_group = predictions[
            mask.to_numpy()
        ]

        n = len(y_group)

        if n == 0:

            r2 = np.nan
            rmse = np.nan
            mae = np.nan
            mape = np.nan

        else:

            r2, rmse, mae, mape = calculate_metrics(
                y_group,
                pred_group
            )

            if n < 2:
                r2 = np.nan

        range_results.append({

            "Model": model_name,

            "Replacement_Group": group,

            "N": n,

            "R2": r2,

            "RMSE_MPa": rmse,

            "MAE_MPa": mae,

            "MAPE_percent": mape
        })

range_results_df = pd.DataFrame(
    range_results
)


range_publication = range_results_df.copy()

range_publication["R²"] = (
    range_publication["R2"]
    .apply(
        lambda x:
        f"{x:.3f}" if pd.notna(x) else "—"
    )
)

range_publication["RMSE (MPa)"] = (
    range_publication["RMSE_MPa"]
    .apply(
        lambda x:
        f"{x:.3f}" if pd.notna(x) else "—"
    )
)

range_publication["MAE (MPa)"] = (
    range_publication["MAE_MPa"]
    .apply(
        lambda x:
        f"{x:.3f}" if pd.notna(x) else "—"
    )
)

range_publication["MAPE (%)"] = (
    range_publication["MAPE_percent"]
    .apply(
        lambda x:
        f"{x:.2f}" if pd.notna(x) else "—"
    )
)

range_publication = range_publication[
    [
        "Model",
        "Replacement_Group",
        "N",
        "R²",
        "RMSE (MPa)",
        "MAE (MPa)",
        "MAPE (%)"
    ]
]

print("\n============================================================")
print("ERROR ANALYSIS BY PUMICE REPLACEMENT RANGE")
print("============================================================")
display(range_publication)


range_results_df.to_csv(
    "Pumice_Replacement_Range_Error_Analysis.csv",
    index=False
)

range_publication.to_csv(
    "Pumice_Replacement_Range_Error_Analysis_Publication.csv",
    index=False
)


# ================================================================
# 18. SAVE OOF + REPLACEMENT INFORMATION
# ================================================================

oof_range_data = df.copy()

for model_name, predictions in oof_predictions.items():
    oof_range_data[
        model_name + "_Predicted"
    ] = predictions

oof_range_data.to_csv(
    "OOF_Predictions_with_Pumice_Replacement_Range.csv",
    index=False
)


# ================================================================
# 19. RMSE BY REPLACEMENT RANGE
# ================================================================

plot_data = (
    range_results_df
    .pivot(
        index="Replacement_Group",
        columns="Model",
        values="RMSE_MPa"
    )
    .reindex(group_order)
)

ax = plot_data.plot(
    kind="bar",
    figsize=(9, 5)
)

ax.set_xlabel(
    "Pumice replacement range"
)

ax.set_ylabel(
    "RMSE (MPa)"
)

ax.set_title(
    "Cross-validation error by pumice replacement range"
)

ax.legend(
    title="Model",
    frameon=False
)

plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    "Pumice_Replacement_Range_RMSE_600dpi.tiff",
    dpi=FIG_DPI,
    bbox_inches="tight"
)

plt.savefig(
    "Pumice_Replacement_Range_RMSE_600dpi.png",
    dpi=FIG_DPI,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 20. SHAP — FIT BEST MODEL TO COMPLETE DATASET FOR INTERPRETATION
# ================================================================
# This model is NOT used to calculate CV performance.
# CV performance remains based on held-out OOF predictions.
# ================================================================

print("\n============================================================")
print("SHAP ANALYSIS")
print("============================================================")

best_model = clone(
    models[best_model_name]
)

best_model.fit(
    X,
    y
)

print(
    "Full-data interpretation model:",
    best_model_name
)


# ================================================================
# 21. SHAP DATA SIZE CONTROL
# ================================================================

if len(X) > MAX_SHAP_OBSERVATIONS:

    rng = np.random.default_rng(
        RANDOM_STATE
    )

    shap_indices = rng.choice(
        len(X),
        size=MAX_SHAP_OBSERVATIONS,
        replace=False
    )

    X_shap = X.iloc[
        shap_indices
    ].copy()

else:

    X_shap = X.copy()


# ================================================================
# 22. SHAP EXPLAINER
# ================================================================

if best_model_name in [
    "RF",
    "GBR",
    "XGB"
]:

    explainer = shap.TreeExplainer(
        best_model
    )

    shap_values = explainer.shap_values(
        X_shap
    )

else:

    # SVR fallback: model-agnostic SHAP.
    # A bounded background set avoids excessive runtime.
    background_size = min(
        MAX_SHAP_BACKGROUND,
        len(X)
    )

    background = shap.sample(
        X,
        background_size,
        random_state=RANDOM_STATE
    )

    explainer = shap.Explainer(
        best_model.predict,
        background
    )

    shap_result = explainer(
        X_shap
    )

    shap_values = shap_result.values


# ================================================================
# 23. SHAP BEESWARM
# ================================================================

plt.figure(
    figsize=(8, 6),
    dpi=FIG_DPI
)

shap.summary_plot(
    shap_values,
    X_shap,
    show=False
)

plt.tight_layout()

plt.savefig(
    "SHAP_Summary_Beeswarm_600dpi.tiff",
    dpi=FIG_DPI,
    bbox_inches="tight"
)

plt.savefig(
    "SHAP_Summary_Beeswarm_600dpi.png",
    dpi=FIG_DPI,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 24. SHAP FEATURE IMPORTANCE
# ================================================================

mean_abs_shap = np.mean(
    np.abs(shap_values),
    axis=0
)

shap_importance_df = pd.DataFrame({

    "Feature": feature_columns,

    "Mean_Absolute_SHAP":
        mean_abs_shap
})

shap_importance_df = (
    shap_importance_df
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

shap_importance_df[
    "Importance_Rank"
] = np.arange(
    1,
    len(shap_importance_df) + 1
)

display(shap_importance_df)

shap_importance_df.to_csv(
    "SHAP_Feature_Importance.csv",
    index=False
)


# ================================================================
# 25. SHAP BAR PLOT
# ================================================================

plt.figure(
    figsize=(8, 5.5),
    dpi=FIG_DPI
)

shap.summary_plot(
    shap_values,
    X_shap,
    plot_type="bar",
    show=False
)

plt.tight_layout()

plt.savefig(
    "SHAP_Feature_Importance_Bar_600dpi.tiff",
    dpi=FIG_DPI,
    bbox_inches="tight"
)

plt.savefig(
    "SHAP_Feature_Importance_Bar_600dpi.png",
    dpi=FIG_DPI,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 26. SHAP DEPENDENCE PLOTS
# ================================================================

for feature in [
    "Water",
    "Sand",
    "Pumice_Aggregate"
]:

    plt.figure(
        figsize=(7, 5.5),
        dpi=FIG_DPI
    )

    shap.dependence_plot(
        feature,
        shap_values,
        X_shap,
        interaction_index="auto",
        show=False
    )

    plt.tight_layout()

    safe_name = (
        feature
        .replace(" ", "_")
        .replace("/", "_")
    )

    plt.savefig(
        f"SHAP_Dependence_{safe_name}_600dpi.tiff",
        dpi=FIG_DPI,
        bbox_inches="tight"
    )

    plt.savefig(
        f"SHAP_Dependence_{safe_name}_600dpi.png",
        dpi=FIG_DPI,
        bbox_inches="tight"
    )

    plt.show()


# ================================================================
# 27. SHAP INTERACTION VALUES
# ================================================================
# Exact Tree SHAP interactions are available for RF, GBR and XGB.
# If SVR happens to be the selected model, this section is skipped
# rather than reporting an inappropriate Tree SHAP interaction result.
# ================================================================

if best_model_name in [
    "RF",
    "GBR",
    "XGB"
]:

    print("\nCalculating SHAP interaction values...")

    interaction_values = (
        explainer.shap_interaction_values(
            X_shap
        )
    )

    mean_abs_interaction = np.mean(
        np.abs(interaction_values),
        axis=0
    )

    interaction_df = pd.DataFrame(
        mean_abs_interaction,
        index=feature_columns,
        columns=feature_columns
    )

    print("\nMean absolute SHAP interaction matrix:")
    display(interaction_df)

    interaction_df.to_csv(
        "SHAP_Interaction_Matrix.csv"
    )


    # Heatmap
    plt.figure(
        figsize=(7, 6),
        dpi=FIG_DPI
    )

    plt.imshow(
        interaction_df.values,
        aspect="auto"
    )

    plt.xticks(
        range(len(feature_columns)),
        feature_columns,
        rotation=45,
        ha="right"
    )

    plt.yticks(
        range(len(feature_columns)),
        feature_columns
    )

    plt.xlabel("Feature")
    plt.ylabel("Feature")

    plt.title(
        "Mean absolute SHAP interaction strength"
    )

    plt.colorbar(
        label="Mean |SHAP interaction|"
    )

    plt.tight_layout()

    plt.savefig(
        "SHAP_Interaction_Heatmap_600dpi.tiff",
        dpi=FIG_DPI,
        bbox_inches="tight"
    )

    plt.savefig(
        "SHAP_Interaction_Heatmap_600dpi.png",
        dpi=FIG_DPI,
        bbox_inches="tight"
    )

    plt.show()


    # Unique pair ranking
    pair_results = []

    for i in range(len(feature_columns)):

        for j in range(i + 1, len(feature_columns)):

            pair_results.append({

                "Feature_1":
                    feature_columns[i],

                "Feature_2":
                    feature_columns[j],

                "Mean_Absolute_Interaction":
                    mean_abs_interaction[i, j]
            })

    pair_interactions_df = (
        pd.DataFrame(pair_results)
        .sort_values(
            "Mean_Absolute_Interaction",
            ascending=False
        )
        .reset_index(drop=True)
    )

    pair_interactions_df["Rank"] = np.arange(
        1,
        len(pair_interactions_df) + 1
    )

    print("\nSHAP interaction pair ranking:")
    display(pair_interactions_df)

    pair_interactions_df.to_csv(
        "SHAP_Interaction_Pair_Ranking.csv",
        index=False
    )


    # Requested interaction dependence plots
    for feature_1, feature_2 in [
        ("Water", "Sand"),
        ("Water", "Pumice_Aggregate"),
        ("Sand", "Pumice_Aggregate")
    ]:

        plt.figure(
            figsize=(7, 5.5),
            dpi=FIG_DPI
        )

        shap.dependence_plot(
            feature_1,
            shap_values,
            X_shap,
            interaction_index=feature_2,
            show=False
        )

        plt.tight_layout()

        safe_pair = (
            f"{feature_1}_x_{feature_2}"
            .replace(" ", "_")
            .replace("/", "_")
        )

        plt.savefig(
            f"SHAP_Interaction_{safe_pair}_600dpi.tiff",
            dpi=FIG_DPI,
            bbox_inches="tight"
        )

        plt.savefig(
            f"SHAP_Interaction_{safe_pair}_600dpi.png",
            dpi=FIG_DPI,
            bbox_inches="tight"
        )

        plt.show()

else:

    print(
        "\nSelected model is SVR; exact Tree SHAP interaction "
        "analysis is not performed."
    )


# ================================================================
# 28. SAVE SHAP VALUES
# ================================================================

shap_values_df = pd.DataFrame(
    shap_values,
    columns=[
        f"SHAP_{x}"
        for x in feature_columns
    ]
)

shap_export_df = pd.concat(
    [
        X_shap.reset_index(drop=True),
        shap_values_df
    ],
    axis=1
)

shap_export_df.to_csv(
    "SHAP_Values.csv",
    index=False
)


# ================================================================
# 29. FINAL SUMMARY
# ================================================================

final_summary = pd.DataFrame({

    "Parameter": [
        "Final observations",
        "Predictors",
        "Cross-validation",
        "Random state",
        "Model selected for SHAP",
        "Mean CV R² of selected model",
        "SD of CV R²",
        "SHAP fitting"
    ],

    "Value": [
        len(df),
        ", ".join(feature_columns),
        f"{N_SPLITS}-fold shuffled CV",
        RANDOM_STATE,
        best_model_name,
        round(best_row["R2_mean"], 6),
        round(best_row["R2_SD"], 6),
        "Complete cleaned dataset; interpretation only"
    ]
})

final_summary.to_csv(
    "Final_Analysis_Summary.csv",
    index=False
)


# ================================================================
# 30. FINAL OUTPUT
# ================================================================

print("\n============================================================")
print("COMPLETE ANALYSIS FINISHED")
print("============================================================")

print("Final observations:", len(df))
print("Selected model:", best_model_name)
print(
    "Mean CV R²:",
    f"{best_row['R2_mean']:.4f}"
)
print(
    "CV R² SD:",
    f"{best_row['R2_SD']:.4f}"
)

print("\nMain files generated:")

for f in sorted(
    glob.glob("*.csv")
    + glob.glob("*.png")
    + glob.glob("*.tiff")
):
    print("✓", f)

print("\n============================================================")
print("MANUSCRIPT-RELEVANT DEFINITIONS")
print("============================================================")
print(
    "1. Model performance = 10-fold held-out out-of-fold predictions."
)
print(
    "2. Replacement-range error analysis = out-of-fold predictions."
)
print(
    "3. SHAP = selected model refitted to the complete cleaned "
    "dataset for interpretation only."
)
print(
    "4. No fixed 131-observation requirement."
)
print(
    "5. Multiple uploaded CSV files are combined and exact "
    "duplicate rows are removed."
)
print("============================================================")
```
